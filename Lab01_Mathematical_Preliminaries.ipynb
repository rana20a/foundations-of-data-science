### 📌 Cell Explanation
This cell shows the **Cumulative Distribution Function (CDF)** of the same dice data.

Unlike the PDF, the CDF shows the probability of getting a value **less than or equal to** a given number:
- At x = 2 → probability ≈ 0.028 (only one outcome: 1+1)
- At x = 12 → probability = 1.0 (100%), because all outcomes are ≤ 12

The CDF is always non-decreasing and reaches exactly 1 at the maximum value. It is useful for answering questions like 'what is the probability of rolling at most 8?'

### Figure 2.6 - Spearman Correlation Coefficient & Pearson Coefficient

```python
np.random.seed(3)
x = np.random.uniform(-3,3,80)
y = np.absolute(x)

rho, pval = stats.spearmanr(x, y)
pearson = stats.pearsonr(x, y)

title = "Spearman coefficient = " + str("%.2f" % rho) + "\nPearson coefficient = " + str("%.2f" % pearson[0])
plt.title(title)

plt.plot(x,y,'ro')
```

### 📌 Cell Explanation
This cell compares **Spearman vs Pearson correlation** on a V-shaped (non-linear) relationship.

- `x` is random uniform between -3 and 3
- `y = |x|` (absolute value) — creates a V-shape

**Results:**
- Pearson ≈ 0 → fails because it only measures **linear** relationships
- Spearman ≈ 0 → also fails because the V-shape is **not monotonic** (it goes down then up)

This demonstrates an important limitation: both coefficients fail to detect non-linear, non-monotonic relationships. Always visualize data before relying on correlation values.

### Figure 2.8 Left

```python
x = np.linspace(0, 1, 300)
line = x*x
plt.plot(x, line, 'r-', linewidth=2.0)
plt.xlabel('Absolute value of Pearson correlation')
plt.ylabel(r'$r^{2}$')
plt.grid()
```

### 📌 Cell Explanation
This cell plots the relationship between the **absolute value of the Pearson coefficient** and its square (R²).

R² represents the **proportion of variance explained** by the model:
- r = 0.7 → R² = 0.49 (only 49% of variance explained!)
- r = 0.9 → R² = 0.81
- r = 1.0 → R² = 1.0 (perfect)

The quadratic growth serves as a **warning against overestimating correlation strength**: a seemingly strong correlation of 0.7 actually explains less than half the variance in the data.

### 📌 Cell Explanation
This cell plots the **minimum Pearson coefficient required for statistical significance** based on sample size, using the approximation: minimum r ≈ 2/√n.

Key insight: as sample size increases, a **smaller correlation** can still be statistically significant.
- With n = 10 samples → need r ≥ 0.63 for significance
- With n = 100 samples → need r ≥ 0.20 for significance

**Lesson:** Sample size is crucial when interpreting correlation. A small correlation in a large dataset may be statistically significant, but that does not mean it is practically meaningful.

### 📌 Cell Explanation
This cell computes the **Autocorrelation Function (ACF)** by shifting the time series by an increasing number of days and calculating the Pearson correlation between the original and shifted versions.

The plot shows **peaks at lags of 7, 14, and 21 days**, confirming the existence of a **weekly cyclical pattern**.

Autocorrelation is a fundamental tool in time series analysis for discovering hidden periodic patterns. It answers the question: 'How similar is the series to a delayed version of itself?'
