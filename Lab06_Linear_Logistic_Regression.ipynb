### Figure 9.1  - Linear Regression graph

```python
# Create a random noise
x1 = np.random.normal(size=200)
xs = np.linspace(x1.min()-1, x1.max()+1, 100)
x = xs
y = 3*(np.random.normal(0, 1, 100)+x)
plt.scatter(x, y , color='red')

#Find the regression line
regression = np.polyfit(x,y,1)
print (regression)
longerX = np.append(x, [5,-5])
plt.plot(longerX, regression[0]*longerX + regression[1], color='black', linewidth='1.5')
plt.xlim(-4, 4.6)
```

### 📌 Cell Explanation

- **Red dots** = actual data points
- **Black line** = best-fit regression line

This is the foundation of predictive modeling: finding the mathematical relationship y = mx + b between input (x) and output (y).

### Figure 9.4 Regression line and outlier

```python
plt.figure(figsize=(12,4.5))

#The right graph without an outlier.
plt.subplot(122)
# Create a random 15 points
np.random.seed(7)
x1 = np.random.normal(size=10)
x = np.linspace(x1.min()-1, x1.max()+1, 15)
y = 3*(np.random.normal(0, 1, 15)+x)
plt.scatter(x, y , color='red')

#Find the regression line
regression = np.polyfit(x,y,1)
longerX = np.append(x, [5,-5]) #this makes the regression line longer
plt.plot(longerX, regression[0]*longerX + regression[1], color='black', linewidth='1.5')
plt.xlim(-3.5,4.5)
plt.ylim(-15, 20)
#The correlation coefficient value r.
r1 = stats.pearsonr(x, y)
print ('r2 =', r1[0])

#The left graph with an outlier
plt.subplot(121)
#plot the 15 points with an outlier.
x1 = np.append(x, [4])
y1 = np.append(y, [-10])
plt.scatter(x1, y1 , color='red')

#Find the regression line
regression = np.polyfit(x1,y1,1)
longerX = np.append(x1, [5,-5]) #this makes the regression line longer
plt.plot(longerX, regression[0]*longerX + regression[1], color='black', linewidth='1.5')
plt.xlim(-3.5,4.5)
plt.ylim(-15, 20)
#The correlation coefficient value r.
r2 = stats.pearsonr(x1, y1)
print ('r1 =', r2[0])
```

### 📌 Cell Explanation
This cell demonstrates how **a single outlier can dramatically affect the regression line**.

- **Right plot** = 15 random points with a strong linear relationship (high Pearson correlation)
- **Left plot** = same data with one outlier at position (4, -10)

The regression line tilts significantly toward the outlier, and the Pearson correlation drops from ~0.92 to ~0.55. This emphasizes the importance of detecting and handling outliers **before** fitting a regression model.

### Figure 9.6 Left

Summary of Fig 9.6
1. Get the points (plot y = x line and make random points around the line by adding noise to the points.)
2. Find the zscore of the points. (x-xmean/xstd, y-ymean/ystd)
3. for angle 0 to 90, find the mean squared error of the line to the points.
4. plot (angle, error)

```python
## Make Points

np.random.seed(7)
x = np.linspace(0,10, 100)
y = np.random.normal(0,1, 100)+x

xmean = np.mean(x)
xstd = np.std(x)
ymean = np.mean(y)
ystd = np.std(y)

## Compute the zscores
xz = []
yz = []
for i, j in zip(x, y):
    xz.append((i-xmean)/float(xstd))
    yz.append((j-ymean)/float(ystd))

plt.figure(figsize=(5,5))

plt.plot(xz, yz, 'bo', markersize=5)
regline = np.polyfit(xz, yz, 1)
plt.plot(xz, [k*regline[0] for k in xz], 'r-', linewidth=3)

plt.grid(linewidth=1.2)
plt.xticks([-2,-1,0,1,2], ['-2','-1','0','1', '2'])
plt.yticks([-2,-1,0,1,2], ['-2','-1','0','1', '2'])
plt.xlabel("x", fontsize=14)
plt.ylabel("y", fontsize=14)
```

### 📌 Cell Explanation
This cell demonstrates regression after **z-score normalization**.

Data is standardized by subtracting the mean and dividing by the standard deviation for both x and y: `z = (value - mean) / std`

After normalization:
- The data is centered at the origin (0, 0)
- Values are measured in units of standard deviation

Z-score normalization is important because it allows **fair comparison between variables with different scales and units**, which is essential before training most machine learning models.

### Figure 9.6 Right

```python
errorlst = []
label = []
#for angle 0 ~ 89 bc when a = 90, it is a vertical line.
for a in range(0,90, 1):
    # slope = y/x
    slope = math.sin(math.radians(a))/float(math.cos(math.radians(a)))
    label.append(slope)
    error = mean_squared_error(yz, [ slope*k for k in xz])
    errorlst.append(error)

#Slope vs. errorlst
plt.figure(figsize=(5,5))
plt.plot(label, errorlst, 'b-', linewidth=2)
plt.plot([1,1],[0,1], '--')
plt.xlim(0,2)
plt.ylim(0,1)
plt.xlabel(r"$w_1$", fontsize=14)
plt.ylabel(r"$J(0,w_1)$", fontsize=14)
plt.grid()
```

### 📌 Cell Explanation
This cell plots the **cost function (MSE)** as a function of the slope parameter.

For each angle from 0 to 89 degrees, the corresponding slope is calculated, and the MSE between actual and predicted values is computed.

The resulting plot shows a **U-shaped (convex) curve** with a minimum at slope = 1 — this is the optimal slope.

This is the core idea of optimization in machine learning: **finding the parameter values that minimize the cost function**.

### 📌 Cell Explanation

Gradient descent works by starting at a random point on this surface and **iteratively moving downhill** toward the minimum. The convex shape guarantees that gradient descent will always find the global minimum.

### Figure 9.8
The tangent line approximates the derivative.

```python
x = np.linspace(-1,2,100)
y = 0.7*np.sin(x)
plt.plot(x,y, linewidth='1.7')

#the tangent line
x0 = 0.0
x1 = 1.0
slope=(0.7*np.sin(x1)-0.7*np.sin(x0))/(x1-x0)
xline = np.linspace(x0-0.5, x1+0.5, 100)
plt.plot(xline, slope*xline, 'r-', linewidth='1.9')

#the points
plt.scatter([x0,x1],[0.7*np.sin(x0), 0.7*np.sin(x1)], s=40, color='black')
plt.text(x0-0.05, 0.7*np.sin(x0)+0.05, 'A')
plt.text(x1, 0.7*np.sin(x1)+0.05, 'B')

#the triangle
plt.plot([x0,x1,x1,x0], [0.7*np.sin(x0), 0.7*np.sin(x1),0.7*np.sin(x0), 0.7*np.sin(x0)], color='black')
plt.text(x0+0.5, 0.7*np.sin(x0)-0.1, r'$\Delta x$', fontsize=14)
plt.text(x1+0.05, 0.7*np.sin(x1)-0.3, r'$\Delta y$', fontsize=14)
plt.text(x0+0.25, 0.7*np.sin(x0)-0.5, r'slope = $\frac{\Delta y}{\Delta x}$', fontsize=14)

plt.xlim(-1,2)

#Hide the x and y axis.
plt.gca().axes.get_xaxis().set_ticks([])
plt.gca().axes.get_yaxis().set_ticks([])
```

### 📌 Cell Explanation
This cell illustrates the concept of a **derivative using a tangent line**.

- **Blue curve** = the function y = 0.7·sin(x)
- **Red line** = the tangent line between points A and B

The slope of this line approximates the derivative (Δy/Δx). As the two points get closer together, the tangent becomes a better approximation of the true derivative.

The derivative tells us the **direction and rate of change** of the function — exactly what gradient descent uses to decide which direction to move to reduce the cost function.

### Figure 9.11 - Bumpy surface

```python
X = np.arange(-2, 2, 0.1)
Y = np.arange(-2, 2, 0.1)
X, Y = np.meshgrid(X, Y)
R = np.sin(X) + np.cos(Y)
R = ((np.cos(X))**4+(np.cos(Y))**4-2*((np.cos(X))**2)*((np.cos(Y))**2))
Z = R
#Equation from http://stackoverflow.com/questions/24045379/matplotlib-3d-surface-of-bumpy-function-does-not-work

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
p = ax.plot_surface(X, Y, Z, rstride=1, cstride=1, cmap=cm.rainbow, linewidth=0)
ax.set_zlim(-1.1,1.1)
ax.set_xlabel(r"$w_0$")
ax.set_ylabel(r"$w_1$")
ax.set_zlabel(r"$J(w_0, w_1)$")
ax.set_xticks(np.arange(-2,2.1,1))
ax.set_yticks(np.arange(-2,2.1,1))
```

### 📌 Cell Explanation
This cell shows a **non-convex (bumpy) cost surface** with multiple local minima and maxima.

Unlike the smooth bowl in Cell 9, this surface has many valleys. This is problematic for gradient descent because the algorithm might get **stuck in a local minimum** instead of finding the global minimum.

This is a common challenge when training complex models like neural networks. Solutions include:
- Random restarts
- Momentum-based optimization
- Learning rate scheduling

### 📌 Cell Explanation
This cell plots the **sigmoid function** — the mathematical core of logistic regression:

σ(x) = 1 / (1 + e⁻ˣ)

The function transforms any input value into a probability between 0 and 1:
- For large positive x → output approaches 1
- For large negative x → output approaches 0
- At x = 0 → output = 0.5

This S-shaped curve is used to convert the linear combination of features into a **probability of belonging to a class**.
